1. Genera un conjunto sintético 2D de dos clases con solapamiento leve.

2. Entrena un SVM lineal con valores de C={0.1,1,10}.

3. Compara la separación obtenida (frontera) y la precisión en test

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import make_blobs
from sklearn.model_selection import train_test_split

# Gener2mos 300 muestras en 2D, 3 centros, leve solapamiento
X, y = make_blobs(n_samples=300, centers=3, cluster_std=3.5, random_state=42)
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)

# Visualización inicial
plt.scatter(X_train[:,0], X_train[:,1], c=y_train, cmap='bwr', alpha=0.7)
plt.title("Datos de entrenamiento")
plt.show()
```

```python
from sklearn.svm import SVC
models = {}
for C in [0.1, 1, 10]:
    svm = SVC(kernel='linear', C=C, random_state=42)
    svm.fit(X_train, y_train)
    models[C] = svm
    print(f"C = {C}: precisión en entrenamiento = {svm.score(X_train, y_train):.2f}, "
          f"precisión en test = {svm.score(X_test, y_test):.2f}")
```

```python
def plot_decision_boundary(model, X, y, ax, title):
    # crear malla
    xx, yy = np.meshgrid(np.linspace(X[:,0].min()-1, X[:,0].max()+1, 200),
                         np.linspace(X[:,1].min()-1, X[:,1].max()+1, 200))
    Z = model.predict(np.c_[xx.ravel(), yy.ravel()]).reshape(xx.shape)
    ax.contourf(xx, yy, Z, alpha=0.2, cmap='bwr')
    ax.scatter(X[:,0], X[:,1], c=y, edgecolor='k', cmap='bwr')
    ax.set_title(title)

# Dibujar para cada C
fig, axes = plt.subplots(1, 3, figsize=(15,4))
for ax, (C, svm) in zip(axes, models.items()):
    plot_decision_boundary(svm, X_test, y_test, ax, f"SVM lineal, C = {C}")
plt.tight_layout()
plt.show()
```

4. Replique el ejercicio anterior usando FashionMNIST

```python
# Cell 1: Cargar Fashion MNIST (dos clases) y reducir a 2D con PCA
import numpy as np
import matplotlib.pyplot as plt
from sklearn.decomposition import PCA
from sklearn.svm import SVC
from sklearn.model_selection import train_test_split
from sklearn.datasets import fetch_openml

mnist = fetch_openml('Fashion-MNIST', version=1, as_frame=False)
X, y = mnist['data'], mnist['target'].astype(int)
```

```python
categories = [
    "T-shirt/top", "Trouser", "Pullover", "Dress", "Coat",
    "Sandal", "Shirt", "Sneaker", "Bag", "Ankle boot"
]
print("Categorías de Fashion MNIST:")
for i, cat in enumerate(categories):
    print(f"{i}: {cat}")

# 3) Plot de ejemplos (un ejemplar por clase)
fig, axes = plt.subplots(2, 5, figsize=(12, 5))
fig.suptitle("Ejemplos de cada categoría", fontsize=16)

for idx, ax in enumerate(axes.flatten()):
    img = X[y == idx][0].reshape(28, 28)
    ax.imshow(img, cmap='gray')
    ax.set_title(categories[idx])
    ax.axis('off')

plt.tight_layout(rect=[0, 0.03, 1, 0.95])
plt.show()
```

```python
# Elegir categoria
mask = (y == 4) | (y == 2)
X_sel, y_sel = X[mask], y[mask]

# División train/val
X_train_full, X_val_full, y_train, y_val = train_test_split(
    X_sel, y_sel, test_size=0.3, random_state=42
)

# PCA a 2 componentes
pca = PCA(n_components=2, random_state=42)
X_train = pca.fit_transform(X_train_full / 255.0)
X_val   = pca.transform(X_val_full   / 255.0)

# Visualizar
plt.figure(figsize=(6,5))
plt.scatter(X_train[:,0], X_train[:,1], c=y_train, cmap='bwr', edgecolor='k', alpha=0.7)
plt.title("Fashion MNIST: T-shirt vs Trouser (PCA 2D)")
plt.xlabel("PC1"); plt.ylabel("PC2")
plt.show()
```

```python
models = {}
for C in [0.1, 1, 10]:
    svm = SVC(kernel='linear', C=C, random_state=42)
    svm.fit(X_train, y_train)
    models[C] = svm
    print(f"C = {C} → train acc: {svm.score(X_train, y_train):.2f}, "
          f"val acc: {svm.score(X_val, y_val):.2f}")
```

```python
def plot_decision_boundary(model, X, y, ax, title):
    xx, yy = np.meshgrid(
        np.linspace(X[:,0].min()-1, X[:,0].max()+1, 200),
        np.linspace(X[:,1].min()-1, X[:,1].max()+1, 200)
    )
    Z = model.predict(np.c_[xx.ravel(), yy.ravel()]).reshape(xx.shape)
    ax.contourf(xx, yy, Z, alpha=0.2, cmap='bwr')
    ax.scatter(X[:,0], X[:,1], c=y, cmap='bwr', edgecolor='k', alpha=0.6)
    ax.set_title(title)
    ax.set_xlabel("PC1"); ax.set_ylabel("PC2")

fig, axes = plt.subplots(1, 3, figsize=(15,4))
for ax, (C, svm) in zip(axes, models.items()):
    plot_decision_boundary(svm, X_val, y_val, ax, f"SVM linear, C={C}")
plt.tight_layout(); plt.show()
```

```python
for C, svm in models.items():
    print(f"Para C = {C}:")
    print(f" - Número de vectores de soporte: {len(svm.support_)}")
    print(f" - Precisión en validación: {svm.score(X_val, y_val):.2f}\n")
```
