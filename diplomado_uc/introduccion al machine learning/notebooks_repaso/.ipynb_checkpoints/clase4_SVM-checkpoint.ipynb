<h3>SVM</h3>
<p>SVM es un algoritmo supervisado para clasificacion y tambien para regresion. Su idea principal es encontrar un "hiperplano" optimo que separe las clases de datos maximizando el margen, es decir, la distancia entre el hiperplano y los puntos mas cercanos de cada clase (llamados vectores de soporte).</p>

<img src="https://upload.wikimedia.org/wikipedia/commons/thumb/7/72/SVM_margin.png/330px-SVM_margin.png" />
<img src="https://ars.els-cdn.com/content/image/3-s2.0-B9780323902779000018-f01-01-9780323902779.jpg" />

<p>El hiperplano optimo es aquel que genera el maximo margen de saerparacion entre las clases.</p>
<p>La idea fundamental detras de SVM es optimizar las distancias, como se aprecia en la imagen de ejemplo</p>
<p>Aquellos puntos que soportan los bordes de mi "off the street" los denominaremos como "support vectors"</p>
<p>Otro concepto clave dentro de las SVM es que si somos estrictos e imponemos que todas las intancias deben estar afuera de la calle y en el lado derecho, esto es llamado "hard margin classification". Estos son dos problemas principales con este concepto; esto solo funciona si los datos son linealmente separables, y que tambien es muy sensible frente a outliers.</p>
<p>Para evitar estos problemas, debemos usar un modelo un poco mas flexible. El objetivo es encontrar un balance global entre mantener la calle lo mas largo posible y limitando la validacion del margen. A esto lo llamaremos "soft margin", el cual permite errores, es tolerante al ruido, tiene una mejor generalizacion y es mejor su uso con datos reales y ruido los cuales no son perfectamente separables.</p>

```python
# usando sckit-learn 
import numpy as np 
import warnings
from sklearn import datasets 
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.svm import LinearSVC
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score

warnings.filterwarnings("ignore")

# importar dataset de iris 
iris = datasets.load_iris(as_frame=True).frame

# separar datos 
X = iris.drop('target',axis=1)
y = iris.target

# datos de entrenamiento y test 
X_train,X_test,y_train,y_test = train_test_split(X,y, random_state=42,test_size=0.2)

# pipeline de entrenamiento 
svm_clf = Pipeline([
    ("scaler", StandardScaler()),
    ("linear_svc", LinearSVC(C=1, loss="hinge"))
])

# entrenar modelo 
svm_clf.fit(X_train,y_train)

# predicts 
y_hat = svm_clf.predict(X_test)
accuracy = accuracy_score(y_hat,y_test)
print(f"accuracy score : {accuracy}")
```

<h3>Clasificacion no lineal</h3>
<p>A pesar de los clasificadores de SVM lineales son eficientes y funcionan sorprendentemente bien en el mayor de los casos, muchos datasets no estan cercanos de ser linealmente separables, para esto podriamos usar tecnicas como "Polynominal Features"</p>

```python
# usando sckit-learn 
import numpy as np 
import warnings
from sklearn import datasets 
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler,PolynomialFeatures
from sklearn.svm import LinearSVC
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score

warnings.filterwarnings("ignore")

# importar dataset de iris 
iris = datasets.load_iris(as_frame=True).frame

# separar datos 
X = iris.drop('target',axis=1)
y = iris.target

# datos de entrenamiento y test 
X_train,X_test,y_train,y_test = train_test_split(X,y, random_state=42,test_size=0.2)

# pipeline de entrenamiento 
svm_clf = Pipeline([
    ("poly", PolynomialFeatures(degree=3)),
    ("scaler", StandardScaler()),
    ("linear_svc", LinearSVC(C=1, loss="hinge"))
])

# entrenar modelo 
svm_clf.fit(X_train,y_train)

# predicts 
y_hat = svm_clf.predict(X_test)
accuracy = accuracy_score(y_hat,y_test)
print(f"accuracy score : {accuracy}")
```
