<h1>K-Nearest Neighbors</h1>

<p>Algoritmo supervisado que puede utilizarse tanto para clasificacion como regresion. Su idea central es predecir el valor o clase de un nuevo punto basandonse en los puntos mas cercanos a el en el conjunto de datos.</p>
<p>Suponiendo que tenemos un conjunto de datos de frutas con variables como peso y color, queremos clasificar una fruta es pera o manza, entonces, KNN busca la K fruta mas parecida en nustro dataset y decide segun la mayoria de votos que tipo de fruta es.</p>

<p>Funcionamiento:</p>
<ul>
    <li>Elije un valor de K: por ej. 3,5 o 7</li>
    <li>Calcula la distancia entre los puntos del conjunto de entrenamiento, la mas comun es la euclidiana</li>
    <li>ordena los vecinos por distancia (menor a mayor)</li>
    <li>Toma los K vecinos mas cercanos</li>
    <li>Clasifica, eligiendo la clase mas frecuente entre estos vecinos, por ej si K=5 y 3 vecinos son "a" y dos "b", el resultado sera "a"</li>
</ul>

<img src="https://medidassimdist.wordpress.com/wp-content/uploads/2019/04/image-2.png"/>

```python
import numpy as np

class KNN:
    # tomamos como parametro principal de la clase la cantidad de vecinos cercanos
    def __init__(self, k=3):
        self.k = k
        # Nota: La convención de sklearn es 'n_neighbors'

    # generamos nuestra funcion de train donde tomamos nuestros datos de entrenamiento
    def fit(self, x_train, y_train):
        # Convertir a numpy arrays si no lo son (útil para compatibilidad con Pandas)
        self.x_train = np.array(x_train)
        self.y_train = np.array(y_train)

    # La distancia es entre dos VECTORES (puntos), no escalares.
    def euclidian_distance(self, x1, x2):
        # x1 y x2 son vectores/arreglos
        # np.sum opera a lo largo de todas las dimensiones de la diferencia al cuadrado
        return np.sqrt(np.sum((x1 - x2) ** 2))

    # generamos la prediccion para cada valor
    def predict(self, x_test):
        # Convertir x_test a numpy array para asegurar la iteración correcta
        x_test_array = np.array(x_test)
        y_pred = [self._predict(x) for x in x_test_array]
        return np.array(y_pred)
    
    def _predict(self, x):
        # calcular distancias entre el punto de prueba 'x' y todos los puntos de entrenamiento
        # Al iterar, 'x_train' ya es un vector, y la función 'euclidian_distance' es correcta ahora
        distancias = [self.euclidian_distance(x, x_train) for x_train in self.x_train]
        
        # obtener índices de los k vecinos más cercanos
        k_indices = np.argsort(distancias)[:self.k] 
        
        # obtener las etiquetas correspondientes
        k_nearest_labels = self.y_train[k_indices]
        
        # devolver la etiqueta más común
        # np.bincount(k_nearest_labels) cuenta las ocurrencias. .argmax() devuelve el índice del valor más alto.
        # En el caso de clasificación, este índice es la propia etiqueta (si las etiquetas son 0, 1, 2...)
        most_common = np.bincount(k_nearest_labels).argmax()
        return most_common

# NOTA ADICIONAL: Si el código usa el atributo .frame de load_iris, 
# x_train y x_test serán objetos Pandas. La función fit corregida con np.array(x_train) 
# asegura el funcionamiento correcto de la clase.
```

<h1>Implementacion basica</h1>
<p>usaremos el dataset de iris para generar un ejemplo</p>

```python
from sklearn.datasets import load_iris
from sklearn.metrics import accuracy_score 

iris = load_iris(as_frame=True).frame
print(iris.info())

# serar datos en features y target 
X = iris.drop(['target'],axis=1)

y = iris.target 

# separar datos en entrenamiento y testo 
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test= train_test_split(X,y,random_state=42, test_size=0.2)

# implementar modelo 
model = KNN(k=10)
model.fit(X_train,y_train)
y_hat = model.predict(X_test)

# revisar metricas 
accuracy = accuracy_score(y_hat,y_test)
print(f"resultado de la prediccion de clasificacion : {accuracy}")
```
