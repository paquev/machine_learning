<h3>Como nos ayuda entonces la entropia a elegir la mejor caracteristica?</h3>
<p>
    <ul>

        <li>cada una de estas particiones tendra su propia entropia</li>
        <li>ahora si, intuitivamente algo como la entropia promedio de las particiones generados podria indicar cuan buena o mala es una caracteristica.</li>
        <li>esta indea es formalizada a traves de una metrica llamada "ganancia de informacion" (informacion gain)</li>
